```python
import sys, os
import anndata as ad
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
cwd = Path(os.getcwd())
datapath = cwd.parent.parent / "data" / "vcc_data" / "adata_Training.h5ad"
adata = ad.read_h5ad(datapath, backed="r")
```

```python
print(adata[0,1])
```

```python
print("\n=== BASIC INFO ===")
print(adata)
print(f"\nShape (cells × genes): {adata.n_obs} × {adata.n_vars}")

print("\n=== OBS (cell metadata) columns ===")
print(list(adata.obs.columns))
print("\nSample obs rows:")
print(adata.obs.head())

print("\n=== VAR (gene metadata) columns ===")
print(list(adata.var.columns))
print("\nSample var rows:")
print(adata.var.head())

print("\n=== MATRIX STATS (X) ===")
X = adata.X

n = adata.n_obs
sample_rows = min(5000, n)  
X_sample = X[:sample_rows]

if hasattr(X_sample, "toarray"):  
    X_block = X_sample.toarray()
elif hasattr(X_sample, "data"):   
    X_block = X_sample.data
else:
    X_block = np.asarray(X_sample)

min_val = float(np.min(X_block))
max_val = float(np.max(X_block))
mean_val = float(np.mean(X_block))

print(f"Type: {type(adata.X)}, dtype: {getattr(adata.X, 'dtype', 'unknown')}")
print(f"(sample of {sample_rows} rows)")
print(f"Min: {min_val:.3g}, Max: {max_val:.3g}, Mean: {mean_val:.3g}")

if len(adata.uns) > 0:
    print("\n=== UNS keys ===")
    print(list(adata.uns.keys()))

if len(adata.layers) > 0:
    print("\n=== LAYERS ===")
    print(list(adata.layers.keys()))
if len(adata.obsm) > 0:
    print("\n=== OBsM (embeddings) ===")
    print(list(adata.obsm.keys()))
```
